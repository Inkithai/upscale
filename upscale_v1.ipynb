### ⚠️ **Important: Uploading Images for Bulk Processing**

For bulk processing, please **compress all your desired images into a single `.zip` file** and then upload *only that ZIP file* using the next cell.

After uploading the ZIP, the notebook will automatically:
1.  Extract the images from the ZIP.
2.  Process each image individually.
3.  Create a new ZIP file containing all the upscaled images for you to download.

**Do NOT upload individual image files.** If you upload multiple individual images, the process will fail with an error. If you only have one image, you can still put it in a ZIP file to use this workflow.

```python
import zipfile
import shutil

print("\nUpload a *single ZIP file* containing images for bulk processing...")
uploaded_files = files.upload()

input_image_paths = [] # List to store paths of all images to be processed

if not uploaded_files:
    print("No file uploaded. Please upload a ZIP file to proceed.")
elif len(uploaded_files) > 1:
    print("Error: Multiple files uploaded. Please upload a *single ZIP file* containing your images.")
else:
    input_zip_filename = list(uploaded_files.keys())[0]
    input_zip_filepath = f"./{input_zip_filename}"

    if not input_zip_filename.lower().endswith('.zip'):
        print(f"Error: The uploaded file '{input_zip_filename}' is not a ZIP file. Please upload a .zip archive.")
    else:
        # Ensure previous extraction directory is clean
        if os.path.exists('input_images'):
            shutil.rmtree('input_images')
        os.makedirs('input_images')

        try:
            # Extract the zip file
            with zipfile.ZipFile(input_zip_filepath, 'r') as zip_ref:
                zip_ref.extractall('input_images')
            print(f"ZIP file '{input_zip_filename}' extracted to 'input_images/'.")

            # Collect all image files from the extracted directory
            for root, _, files_in_dir in os.walk('input_images'):
                for file_name in files_in_dir:
                    # Filter for common image extensions
                    if file_name.lower().endswith(('.png', '.jpg', '.jpeg', '.webp')):
                        input_image_paths.append(os.path.join(root, file_name))

            if not input_image_paths:
                print("No supported image files found in the uploaded ZIP. Supported formats: PNG, JPG, JPEG, WEBP.")
            else:
                print(f"Found {len(input_image_paths)} images in the ZIP for processing.")

        except zipfile.BadZipFile:
            print(f"Error: The file '{input_zip_filename}' is not a valid ZIP file. Please check your archive.")
        except Exception as e:
            print(f"An unexpected error occurred during ZIP extraction: {e}")

# At this point, input_image_paths will contain the list of images to process or be empty if an error occurred.
# The next step would be to iterate through input_image_paths in the upscaling cell.
```

```python
if not input_image_paths: # Changed from input_image_path is None
    print("Skipping AI model loading as no valid images were uploaded.")
else:
    print("\nDownloading and configuring AI model...")
    model_name = 'RealESRGAN_x4plus' # Using the default 4x model as requested

    # Initialize upsampler to None before any checks or assignments
    upsampler = None

    # Define model weight path
    model_weight_path = f'weights/{model_name}.pth'
    if not os.path.exists('weights'):
        os.makedirs('weights')

    # Download model if not already present
    if not os.path.exists(model_weight_path):
        print(f"Downloading {model_name}.pth...")
        os.system(f'wget https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/{model_name}.pth -P weights --quiet')
        if not os.path.exists(model_weight_path):
            print(f"Error: Failed to download {model_name}.pth. Please check your internet connection or model availability.")
            # upsampler already None, so no explicit assignment needed here unless there's more complex error handling
        else:
            print(f"Downloaded {model_name}.pth.")
    else:
        print(f"Model '{model_name}.pth' already exists locally. Skipping download.")

    # Initialize the model architecture based on the selected model name
    if upsampler is not None or os.path.exists(model_weight_path): # Only proceed if model downloaded/exists
        if model_name == 'RealESRGAN_x4plus':
            model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=4)
        elif model_name == 'RealESRGAN_x4plus_anime_6B':
            model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=6, num_grow_ch=32, scale=4)
        elif model_name == 'realesr-general-x4v3':
            model = SRVGGNetCompact(num_in_ch=3, num_out_ch=3, num_feat=64, num_conv=32, upscale=4, act_type='prelu')
        else:
            raise ValueError(f"Unsupported model name: {model_name}")

        # Initialize RealESRGANer
        upsampler = RealESRGANer(
            scale=4, # Default upscale factor 4x
            model_path=model_weight_path,
            model=model,
            tile=500, # Process image in tiles to prevent GPU memory errors for large images
            tile_pad=10, # Pad for tiling
            pre_pad=0,
            half=True if torch.cuda.is_available() else False, # Use half precision (FP16) on GPU if available
            gpu_id=0 if torch.cuda.is_available() else None # Use GPU 0 if available, else no specific GPU ID
        )
        # Move model to the selected device (CPU/GPU) and set to evaluation mode
        upsampler.model.eval().to(device)
        print("AI model loaded successfully.")
    else:
        print("AI model loading skipped due to previous download error or model file not found.")
```

```python
output_dir = 'upscaled_images'
if os.path.exists(output_dir):
    shutil.rmtree(output_dir)
os.makedirs(output_dir)

if not input_image_paths or 'upsampler' not in locals() or upsampler is None:
    print("Skipping AI upscaling as no valid images were uploaded or model was not loaded.")
else:
    print(f"\nRunning AI upscaling for {len(input_image_paths)} images...")
    processed_output_paths = []

    for i, current_input_image_path in enumerate(input_image_paths):
        print(f"\nProcessing image {i+1}/{len(input_image_paths)}: {os.path.basename(current_input_image_path)}")
        try:
            # Read image directly from the extracted file path
            img_pil = PILImage.open(current_input_image_path).convert('RGB') # Ensure RGB format
            img_np = np.array(img_pil) # Convert PIL image to NumPy array

            # RealESRGAN expects BGR format for input via OpenCV, so convert if needed
            img_bgr = cv2.cvtColor(img_np, cv2.COLOR_RGB2BGR)

            with torch.no_grad(): # Disable gradient calculations for inference to save memory and speed up
                output_bgr_np, _ = upsampler.enhance(img_bgr, outscale=4) # Perform upscaling

            # Convert the output from BGR back to RGB for display and saving with PIL
            output_rgb_np = cv2.cvtColor(output_bgr_np, cv2.COLOR_BGR2RGB)
            final_upscaled_pil = PILImage.fromarray(output_rgb_np)

            # Generate output filename
            base_name = os.path.splitext(os.path.basename(current_input_image_path))[0]
            output_filename_base = f"{base_name}_upscaled_final"
            current_output_filepath = os.path.join(output_dir, f"{output_filename_base}.jpg")

            # Save the upscaled image (without size modification yet)
            final_upscaled_pil.save(current_output_filepath, format='JPEG', quality=95)
            processed_output_paths.append(current_output_filepath)
            print(f"Upscaled image saved to: {current_output_filepath}")

        except torch.cuda.OutOfMemoryError:
            print(f"\nError: Ran out of GPU memory for {os.path.basename(current_input_image_path)}. Skipping this image.")
            print("Consider adjusting the 'tile' parameter in RealESRGANer initialization to a smaller value if you encounter this frequently.")
        except Exception as e:
            print(f"\nAn unexpected error occurred during upscaling or processing for {os.path.basename(current_input_image_path)}: {e}")
            print("Please ensure the image is valid and try again. Skipping this image.")

    print("\nAll available images processed.")
    # Store the list of processed output paths for the next cell
    output_filepaths_for_download = processed_output_paths
```

```python
if not output_filepaths_for_download:
    print("Skipping file size check as no upscaled images were processed.")
else:
    print(f"\nChecking file size for {len(output_filepaths_for_download)} upscaled images...")
    target_min_size_bytes = 4.1 * 1024 * 1024  # 4.1 MB in bytes

    # Temporary list to hold paths of images that have been finalized (potentially resized)
    final_output_filepaths = []

    for i, original_output_path in enumerate(output_filepaths_for_download):
        print(f"\nProcessing size for image {i+1}/{len(output_filepaths_for_download)}: {os.path.basename(original_output_path)}")

        try:
            # Load the upscaled image from its path
            current_image_for_sizing = PILImage.open(original_output_path).convert('RGB')

            img_buffer = io.BytesIO()
            # Save initially to check size, use quality=100, optimize=False, subsampling=0 for max quality
            current_image_for_sizing.save(img_buffer, format='JPEG', quality=100, optimize=False, subsampling=0)
            current_file_size = img_buffer.tell()
            img_buffer.seek(0)  # Reset buffer position for reading

            if current_file_size < target_min_size_bytes:
                print(f"Initial file size ({current_file_size / (1024 * 1024):.2f} MB) is less than 4.1 MB. Resizing to increase size...")
                iteration = 0
                while current_file_size < target_min_size_bytes:
                    iteration += 1
                    # Limit iterations to prevent infinite loops for extremely small images or other issues
                    if iteration > 20: # Arbitrary limit, adjust as needed
                        print(f"Warning: Stopped resizing after {iteration-1} iterations for {os.path.basename(original_output_path)}. Could not reach target size.")
                        break

                    print(f"Iteration {iteration}: Current size {current_file_size / (1024 * 1024):.2f} MB. Increasing dimensions...")

                    # Increase dimensions by a small factor (e.g., 5-10%)
                    new_width = int(current_image_for_sizing.width * 1.05)
                    new_height = int(current_image_for_sizing.height * 1.05)

                    # Ensure dimensions actually increase to avoid infinite loop with integer rounding
                    if new_width <= current_image_for_sizing.width and new_height <= current_image_for_sizing.height:
                        new_width = current_image_for_sizing.width + 1
                        new_height = current_image_for_sizing.height + 1

                    current_image_for_sizing = current_image_for_sizing.resize((new_width, new_height), PILImage.LANCZOS)

                    img_buffer = io.BytesIO()
                    current_image_for_sizing.save(img_buffer, format='JPEG', quality=100, optimize=False, subsampling=0)
                    current_file_size = img_buffer.tell()
                    img_buffer.seek(0)  # Reset buffer position
                    # time.sleep(0.05) # Small delay for better progress message readability (optional)

                print(f"Final image dimensions after resizing: {current_image_for_sizing.width}x{current_image_for_sizing.height}")
            else:
                print(f"Initial file size ({current_file_size / (1024 * 1024):.2f} MB) already meets the 4.1 MB requirement. No additional resizing needed.")

            # Overwrite the original upscaled image file with the (potentially resized) version
            with open(original_output_path, 'wb') as f:
                f.write(img_buffer.getvalue())

            final_file_size_mb = current_file_size / (1024 * 1024)
            print(f"Final file size for {os.path.basename(original_output_path)}: {final_file_size_mb:.2f} MB")
            final_output_filepaths.append(original_output_path) # Add to the list of successfully processed files

        except Exception as e:
            print(f"Error processing size for {os.path.basename(original_output_path)}: {e}")
            print("Skipping this image for final download.")

    # Update output_filepaths_for_download to only include files that passed the size check and were saved
    output_filepaths_for_download = final_output_filepaths
    print("\nAll images processed for size requirements.")
```
